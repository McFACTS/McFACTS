```python
import numpy as np
import matplotlib.pyplot as plt

from mcfacts.modules.gas_hardening import primary_drag_force_components_base, secondary_drag_force_components_base

mach_range = np.linspace(0, 8, 100)

primary_drag_force_components = np.vectorize(primary_drag_force_components_base)
secondary_drag_force_components = np.vectorize(secondary_drag_force_components_base)

primary = primary_drag_force_components(mach_range)
secondary = secondary_drag_force_components(mach_range)

values = np.abs(secondary[1] / primary[1])

plt.plot(mach_range, values)
```
